# Water Jug Problem using DFS

This notebook implements the Water Jug Problem using depth-first search (DFS) and prints the rule applied at each step in the solution path.

```python
capA = 4
capB = 3
target = 2

visited_states = set()

def solve_dfs(a, b):
    
    if (a, b) in visited_states:
        return False

    visited_states.add((a, b))

    print("Current State:", (a, b))

    if a == target or b == target:
        print("\nTarget Achieved!")
        return True

    #1 Fill Jug A
    if solve_dfs(capA, b):
        print("Applied Rule: Fill Jug A")
        return True

    #2 Fill Jug B
    if solve_dfs(a, capB):
        print("Applied Rule: Fill Jug B")
        return True

    #3 Empty Jug A
    if solve_dfs(0, b):
        print("Applied Rule: Empty Jug A")
        return True

    #4 Empty Jug B
    if solve_dfs(a, 0):
        print("Applied Rule: Empty Jug B")
        return True

    #5 Pour A to B
    pour = min(a, capB - b)
    if solve_dfs(a - pour, b + pour):
        print("Applied Rule: Pour A -> B")
        return True

    #6 Pour B to A
    pour = min(b, capA - a)
    if solve_dfs(a + pour, b - pour):
        print("Applied Rule: Pour B -> A")
        return True

    return False

print("Starting DFS...\n")
solve_dfs(0, 0)
```
